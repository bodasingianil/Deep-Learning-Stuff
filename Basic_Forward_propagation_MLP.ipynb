### Understanding Forward Propagation



```
Input Layer        Hidden Layer              Output Layer
(2 neurons)        (4 neurons)               (1 neuron)

   x1  ● ─────┐      ● h1 ─────┐
              ├────▶ ● h2 ─────┼────▶ ● y
   x2  ● ─────┘      ● h3 ─────┘
                     ● h4

```

````python
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])

# X.shape = (4, 2)

# 4 rows → 4 training samples

# 2 columns → 2 input features = Number of Neurons in input layer
````

````python
# Initialize weights and biases
np.random.seed(1)
input_dim = 2 #X.shape[1]
hidden_dim = 4
output_dim = 1
````

````python
W1 = np.random.randn(input_dim, hidden_dim) # (2,4) = (rows, cols)
# Here, rand (2,4) provides 2*3 = 6 equally distributed values.
W1
````

````python
b1 = np.zeros((1, hidden_dim))
b1
````

````python
W2 = np.random.randn(hidden_dim, output_dim)
b2 = np.zeros((1, output_dim))
````

````python
# Sigmoid activation
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
````

````python
# Forward Propagation
def forward_propagation(X, W1, b1, W2, b2):
  z1 = np.dot(X, W1) + b1
  A1 = sigmoid(z1)

  z2 = np.dot(A1, W2) + b2
  A2 = sigmoid(z2)
  return A2
````

````python
output = forward_propagation(X, W1, b1, W2, b2)
````

````python
print("Output from Forward Propagation:\n", output)
````
